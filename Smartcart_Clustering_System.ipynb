## Feature selection for clustering

```python
features = ['Income', 'Total_Spending', 'Total_Purchases']

X = df[features]
X.head()
```

```python
# Handle Missing Values

X.isnull().sum()
```

```python
X['Income'] = X['Income'].fillna(X['Income'].median())
```

## STANDARDIZATION

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=features)

X_scaled.head()
```

## Elbow Method

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
    
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(1,11),wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Numbers of Clusters")
plt.ylabel("WCSS")
plt.xticks(range(1,11))
plt.grid(True)

plt.show()
```

```python
# KMeans with 4
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

clusters = kmeans.fit_predict(X_scaled)
df['Cluster'] = clusters
df['Cluster'].value_counts()
```

```python
df['Cluster'] = kmeans.labels_
```

```python
df['Cluster'].value_counts().sort_index()
```

```python
df.groupby('Cluster')[features].mean().round(2)
```

```python
sns.scatterplot(data=df, x='Income', y='Total_Spending', hue='Cluster', palette='viridis')

plt.title('Customer Segments: Income vs Spending')
plt.xlabel('Income')
plt.ylabel('Total Spending')
plt.show()
```

# Model evaluation

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X_scaled, df['Cluster'])
print("Silhouette Score:", score)
```

## Hierarchical Clustering

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=4, linkage='ward')

hierarchial_clusters = hc.fit_predict(X_scaled)

df['Hierarchial_Clusters'] = hierarchial_clusters
```

```python
df['Hierarchial_Clusters'].value_counts().sort_index()
```

```python
df.groupby('Hierarchial_Clusters')[features].mean().round(2)
```

## Model evaluation

```python
from sklearn.metrics import silhouette_score

hc_score = silhouette_score(X_scaled, df['Hierarchial_Clusters'])
print("Hierarchial Silhouette Score:", hc_score)
```

```python
# Comparison Table

comparison = pd.DataFrame({
    'Algorithm': ['K-Means','Hierarchial'],
    'Silhouette Score': [score, hc_score]
})

comparison
```
